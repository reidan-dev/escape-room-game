# Escape Room Phonebook Filter

Run the cells below **one at a time, in order**. Each cell applies one clue from `filter_rules.yaml` to the `current` DataFrame and shows what's left, so you can manually check the elimination at every step.

Re-run the setup cell to reset back to the full list of 1000 names.

```python
import pandas as pd

VOWELS = set("aeiou")


def vowel_count(s: str) -> int:
    return sum(1 for c in s.lower() if c in VOWELS)


def common_letters_ordered(a: str, b: str):
    """Letters common to both strings, in the order they appear in `a`,
    plus whether that same relative order holds in `b`."""
    a, b = a.lower(), b.lower()
    common = sorted(set(a) & set(b), key=a.index)
    positions_in_b = [b.index(c) for c in common]
    in_order = all(positions_in_b[i] <= positions_in_b[i + 1] for i in range(len(positions_in_b) - 1))
    return common, in_order


df = pd.read_csv("../data/phonebook.csv", dtype={"phone": str})
# "special" rows (hotline numbers, no name) aren’t name-elimination candidates.
df = df[df["role"] != "special"].reset_index(drop=True)
current = df.copy()
print(f"Loaded {len(df)} entries.")
current.drop(columns=["role"])  # role hidden while filtering, checked at the end
```

## Rule 1
> First name is not 3 letters long.

```python
before = len(current)
mask = current["first_name"].str.len() != 3
current = current[mask]
print(f"{before} -> {len(current)} remaining.")
current.drop(columns=["role"])
```

## Rule 2
> First name starts with a consonant.

```python
before = len(current)
mask = ~current["first_name"].str[0].str.lower().isin(VOWELS)
current = current[mask]
print(f"{before} -> {len(current)} remaining.")
current.drop(columns=["role"])
```

## Rule 3
> Last name has 5 or more letters.

```python
before = len(current)
mask = current["last_name"].str.len() >= 5
current = current[mask]
print(f"{before} -> {len(current)} remaining.")
current.drop(columns=["role"])
```

## Rule 4
> Last name does not end in a vowel.

```python
before = len(current)
mask = ~current["last_name"].str[-1].str.lower().isin(VOWELS)
current = current[mask]
print(f"{before} -> {len(current)} remaining.")
current.drop(columns=["role"])
```

## Rule 5
> First name contains the letter A.

```python
before = len(current)
mask = current["first_name"].str.lower().str.contains("a")
current = current[mask]
print(f"{before} -> {len(current)} remaining.")
current.drop(columns=["role"])
```

## Rule 6
> First and last names do not start with the same letter.

```python
before = len(current)
mask = current["first_name"].str[0].str.lower() != current["last_name"].str[0].str.lower()
current = current[mask]
print(f"{before} -> {len(current)} remaining.")
current.drop(columns=["role"])
```

## Rule 7
> First name has more letters than the last name.

```python
before = len(current)
mask = current["first_name"].str.len() > current["last_name"].str.len()
current = current[mask]
print(f"{before} -> {len(current)} remaining.")
current.drop(columns=["role"])
```

## Rule 8
> First and last names have the same number of vowels.

```python
before = len(current)
mask = current["first_name"].apply(vowel_count) == current["last_name"].apply(vowel_count)
current = current[mask]
print(f"{before} -> {len(current)} remaining.")
current.drop(columns=["role"])
```

## Rule 9
> First and last names share at least one letter.

```python
before = len(current)
mask = current.apply(lambda row: len(set(row["first_name"].lower()) & set(row["last_name"].lower())) > 0, axis=1)
current = current[mask]
print(f"{before} -> {len(current)} remaining.")
current.drop(columns=["role"])
```

## Rule 10
> The first and last names have exactly two letters in common, and those letters appear in the same order in both names.

```python
before = len(current)
mask = current.apply(lambda row: (lambda c, ordered: len(c) == 2 and ordered)(*common_letters_ordered(row["first_name"], row["last_name"])), axis=1)
current = current[mask]
print(f"{before} -> {len(current)} remaining.")
current.drop(columns=["role"])
```

## Final Answer

```python
assert len(current) == 1, f"expected exactly 1 survivor, got {len(current)}"
assert current["role"].iloc[0] == "target", f"survivor's role is {current['role'].iloc[0]!r}, not 'target'"
print("Matches the target row. Puzzle checks out.")
current.drop(columns=["role"])
```
